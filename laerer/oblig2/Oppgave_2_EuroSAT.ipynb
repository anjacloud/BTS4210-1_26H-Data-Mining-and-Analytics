```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

## Del 1 – Datapreparering og utforskning

```python
DATA = "../Data/eurosat.npz" # Velg gjerne 64x64 hvis dere har tid til å trene på det. 
klasser = ["Åker (ettårig)", "Skog", "Urtevegetasjon", "Motorvei", "Industri",
           "Beite", "Åker (flerårig)", "Boligområde", "Elv", "Sjø/innsjø"]

data = np.load(DATA)
X_train, y_train = data["train_images"], data["train_labels"].ravel()
X_val, y_val = data["val_images"], data["val_labels"].ravel()
X_test, y_test = data["test_images"], data["test_labels"].ravel()
print("Trening:", X_train.shape, " validering:", X_val.shape, " test:", X_test.shape)
```

```python

fig, ax = plt.subplots(len(klasser), 6, figsize=(9, 1.6 * len(klasser)))
for k in range(len(klasser)):
    for j, i in enumerate(np.where(y_train == k)[0][:6]):
        ax[k, j].imshow(X_train[i]); ax[k, j].axis("off")
    ax[k, 0].set_title(klasser[k], fontsize=9, loc="left")
plt.tight_layout(); plt.show()
```
